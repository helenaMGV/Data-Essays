```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pytrends.request import TrendReq

# Configuración
pytrends = TrendReq(hl='es-MX', tz=360)

# Términos a comparar
terminos_salud = ['ansiedad', 'depresion', 'terapia']
terminos_consumo = ['black friday', 'buen fin', 'amazon']

# Descarga de datos
pytrends.build_payload(
    terminos_salud + terminos_consumo,
    timeframe='2019-01-01 2024-12-31',
    geo='MX'
)

df = pytrends.interest_over_time()
df = df.drop(columns=['isPartial'])

print(df.head())
df.to_csv('datos/google_trends_raw.csv')
```
